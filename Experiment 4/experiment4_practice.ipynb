```python
import re
import nltk
import string
import math
import pandas as pd 
from collections import Counter
import spacy

nlp = spacy.load('en_core_web_sm')
```

```python
with open('4.1_4.2_input.txt', 'r', encoding='utf-8') as f:
    text = f.read()

doc = nlp(text)
```

```python
words = [token.text.lower() for token in doc if token.is_alpha()]
tf = Counter(words)

df_tf_toolkit = pd.DataFrame(tf.most_common(), columns=['Term', 'Frequency'])
df_tf_toolkit.to_csv('4.1_tf.csv', index = False)

print("Top 10 most frequent terms (Toolkit):")
print(df_tf_toolkit.head(10))

print("\nNamed Entities (Sample):")
seen_entities = set()
for ent in doc.ents:
    if ent.text not in seen_entities:
        seen_entities.add(ent.text)
```
